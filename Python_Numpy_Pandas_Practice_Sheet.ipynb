```python
35.
import numpy as np

np.random.seed(42)
marks = np.random.randint(20, 101, (12, 4))

avg = marks.mean(axis=1)
mean = marks.mean(axis=0)
std = marks.std(axis=0)

print("Marks:\n", marks)
print("Student averages:", avg)
print("Test means:", mean)
print("Test standard deviations:", std)
print("Pass all four:", np.sum(np.all(marks >= 40, axis=1)))
print("Students with average >= 75:", np.where(avg >= 75)[0] + 1)
print("Test with greatest variation:", np.argmax(std) + 1)
```

```python
36.
import numpy as np

a = np.arange(21, 46).reshape(5, 5)
b = a[1:4, 1:4]
c = b.T
d = b @ c

row_sums = d.sum(axis=1)
maximum = d.max()
positions = np.argwhere(d == maximum)

print("Original matrix:\n", a)
print("Central matrix:\n", b)
print("Transpose:\n", c)
print("Product:\n", d)
print("Row sums:", row_sums)
print("Maximum:", maximum)
print("Positions (zero-based):\n", positions)
```

```python
37.
import numpy as np

qty = np.array([
    [10, 15, 20, 8],
    [12, 18, 16, 10],
    [14, 20, 25, 12],
    [18, 22, 20, 15],
    [16, 25, 18, 20],
    [20, 18, 30, 16]
])

price = np.array([150, 240, 90, 320])

daily_revenue = (qty * price).sum(axis=1)
product_qty = qty.sum(axis=0)
product_revenue = (qty * price).sum(axis=0)

new_price = np.where(product_qty > 100, price * 0.90, price)
new_revenue = (qty * new_price).sum()

print("Daily revenue:", daily_revenue)
print("Product quantities:", product_qty)
print("Revenue per product:", product_revenue)
print("Highest revenue day:", daily_revenue.argmax() + 1)
print("Best-selling product:", product_qty.argmax() + 1)
print("New prices:", new_price)
print("Revenue after discount:", new_revenue)
print("Original prices:", price)
```

```python
38.
import numpy as np

np.random.seed(42)
data = np.random.normal(50, 12, 100)

cleaned = np.where(data < 30, 30,
                   np.where(data > 70, 70, data))

print("Values outside 30-70:", np.sum((data < 30) | (data > 70)))

print("Original statistics:")
print("Mean:", data.mean())
print("Variance:", data.var())
print("Standard deviation:", data.std())
print("Minimum:", data.min())
print("Maximum:", data.max())

print("\nCleaned statistics:")
print("Mean:", cleaned.mean())
print("Variance:", cleaned.var())
print("Standard deviation:", cleaned.std())
print("Minimum:", cleaned.min())
print("Maximum:", cleaned.max())
```

```python
39.
import pandas as pd

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

result = branch_a + branch_b + branch_c

zero_result = (
    branch_a.add(0, fill_value=0)
    + branch_b.add(0, fill_value=0)
    + branch_c.add(0, fill_value=0)
)

print("Ordinary addition:\n", result)
print("Treat missing records as zero:\n", zero_result)

common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("Common quarters:", common.tolist())
print("Combined common earnings:", result.loc[common].sum())
```

```python
40.
import pandas as pd

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores, index=required)

print("Original Series:\n", s)
print("Missing records:\n", s[s.isnull()])

filtered = s[s.notnull() & (s >= 70)]
print("Scores >= 70:\n", filtered)

d = filtered.to_dict()
new_s = pd.Series(d)

print("Dictionary:", d)
print("Reconstructed Series:\n", new_s)
print("Values match:", filtered.equals(new_s))
```
