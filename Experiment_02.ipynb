## Obejects,Identity (is) vs Equality(==)

proble 3.1 :
== compares values. is checks whether two variables refer to the exact same object.

```python
# problem 3.2:
#Expected output: True followed by False.
a = [1,2,3]
b = [1,2,3]
print(a == b)
print(a is b)
```

```python
# problem 3.3:
#Python commonly reuses some small integer objects, so is can be True for small
#integers. Identity for larger integers is implementation-dependent;
# use == for numeric comparison.
x = 100
y = 100
print(x == y)
print(x is y)

p = 1000
q = 1000
print(p == q)
print(p is q)
```

## Conditions-if/elif/else,modulo(%)

```python
# problem 6.1:
i = 1
while i<20:
  if i%2 == 0:
    print(i)

  i +=1
```

```python
# problem 6.2:

for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#problem 6.3:
# Expected output: 0, 11, -8, 3, 14
for i in range(0,5):
  if i%3 ==0:
    print(i)
  elif i%3 == 1:
    print(i+10)
  else:
    print(i-10)
```
